```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import datetime as dt
```

```python
CHEMIN = "./"
```

```python
data = pd.read_csv(CHEMIN + "daily_ozone_and_covariates.csv")
data.head()
```

```python
data.shape
```

```python
data["day"] = pd.to_datetime(data["day"])
```

```python
data["week"] = data["day"].dt.to_period("W")
```

```python
weekly_data = data.groupby(["idSensor", "week"]).agg({"maxTemp": ['max'], 
                                        "minTemp": ['min'], 
                                        "rainSum": ['sum'], 
                                        "windSpeed": ['max'],
                                       "maxOzone": ['max']}).reset_index()
```

```python
weekly_data.columns = ["idSensor", "week", "maxTemp", "minTemp", "rainSum", "windSpeed", "maxOzone"]
```

```python
weekly_data.head()
```

```python
weekly_data.to_csv(CHEMIN + "weekly_ozone_and_covariates.csv", header=True, index=False)
```
